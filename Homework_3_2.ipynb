Важливо навчитись висловлювати свої гіпотези щодо даних, і з часом та досвідом вони будуть ставати щоразу глибше і точніше. Весь аналіз даних будується на тому, що ми перевіряємо певні гіпотези та робимо висновки і рекоемндації на основі своїх досліджень.

```python
df[df['City'].isin(['Yangon', 'Mandalay'])].groupby('City')['gross income'].mean()
```

В Mandalay більше середнє gross income, ніж у Yangon.
Причина не у кількості продажів, адже у Yangon вона більша.

Причина може бути у вартості за одну одиницю, у Mandalay в середньому Unit price вище. Тобто клієнти купляють дорожчі продукти. Можливо через більший рівень доходу в жителів міста. В такому випадку у Yangon потрібно залучати більше покупців.

Також у Mandalay трохи більша середня кількість товарів на одного клієнта.
Відповідно і більший середній чек.

Ще може залежати від статі споживачів (в Mandalay відсоток жінок-покупців переважає Yangon).

Розв'язок:

```python
mean_Yangon = df[df.City=='Yangon']['gross income'].mean()
mean_Mandalay = df[df.City=='Mandalay']['gross income'].mean()
```

```python
mean_Yangon, mean_Mandalay
```

Середнє значення gross income у Yangon становить 14.87, тоді як у Mandalay — 15.23. Це означає, що середній дохід від продажів у супермаркетах у Mandalay трохи вищий, ніж у Yangon. Три гіпотези, чому це може бути так:

1. **Відмінності в споживацькому попиті**: Можливо, у Mandalay вищий попит на деякі товари, що приносять більший прибуток. Це може бути через відмінності в демографії, культурних звичках або доступності альтернативних торгових точок. Аби це зʼясувати - маємо порівняти продажі різних груп товарів в містах і знайти, яка група користується більшим попитом в Mandalay. Можливо варто збільшити поставки цих товарів в цьому місті, аби заробляти більше.

2. **Різниця в пробутковості товарів через локальні правила**: Можливо у різних містах різна прибутковість одних і тих самих товарів, і тому за однакового попиту, середня прибутковість в Mandalay - вища. Аби це зʼясувати можемо дослідити маржу на товари в двох містах.

3. **Різноманітність та доступність асортименту**: У супермаркетах у Mandalay може бути більший або кращий асортимент товарів, що приваблює більше покупців та сприяє збільшенню середнього доходу. Це також може бути пов'язано з ефективнішою маркетинговою стратегією або кращим управлінням запасами. Аби це дослідити в ідеалі мати доступ до товарів, які поставляються в кожному з міст. Можна також переглянути, чи по факту покупок в Mandalay купують більш різноманітні товари або просто інші, ніж в Yangon.

Наведені гіпотези надалі можуть бути підтверджені або спростовані через додатковий аналіз даних, такий як вивчення патернів покупок, аналіз цін та маркетингових активностей, а також оцінка зовнішніх економічних і соціальних факторів у цих містах.

9. Врахуовуючи висновок в попередньому завданні, дослідимо як продаються в цих містах різні категорії товарів. Виведіть відсотковий розподіл кількості інвойсів по категоріям товарів `Product line` в містах `Yangon` та `Mandalay`.

```python
df[df['City'].isin(['Yangon', 'Mandalay'])].groupby('City')['Product line'].value_counts(normalize=True)
```

Розв'язок:

```python
df[df.City=='Yangon']['Product line'].value_counts(normalize=True)
```

```python
df[df.City=='Mandalay']['Product line'].value_counts(normalize=True)
```

```python
df_sel_cols = df[df['City'].isin(['Yangon', 'Mandalay'])]

counts = df_sel_cols.groupby(['City', 'Product line']).size().unstack(fill_value=0)

percentages = counts.div(counts.sum(axis=1), axis=0)

percentages
```
